```python
from typing import Annotated,Literal
from typing import TypedDict
from langgraph.graph import StateGraph
from langgraph.constants import START,END
from langgraph.types import Command
class Comman_main(TypedDict):
    score:str
    way:str
def scores(state:Comman_main):
    score=int(input("请输入分数"))
    return {"score":score}
def fail_node(state:Comman_main):
    print(f"接收到分数:{state['score']}")
    return {"way":"失败"}
def scuss_node(state:Comman_main):
    print(f"接收到分数:{state['score']}")
    return {"way":"成功"}
def router(state:Comman_main)->Literal["fail","scuss"]:
    score=state["score"]
    if score>=60:
        return "scuss"
    return "fail"
build_graph=StateGraph(Comman_main)
build_graph.add_node("score",scores)
build_graph.add_node("scuss",scuss_node)
build_graph.add_node("fail",fail_node)
build_graph.add_edge(START,"score")
build_graph.add_conditional_edges("score",router)
build_graph.add_edge("fail",END)
build_graph.add_edge("scuss",END)
result_graph=build_graph.compile()
result=result_graph.invoke({"score":0,"way":"fail"})
print(result)

```
